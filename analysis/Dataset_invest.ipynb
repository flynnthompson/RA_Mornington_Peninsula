# **Dataset Investigation for Mornington Peninsula RA Work**

**Maximum Temperature DS - 1910-2025**

```python
import xarray as xr

base_path = '/g/data/zv2/agcd/v1-0-4'

ds_tmax = xr.open_mfdataset(
    f'{base_path}/tmax/mean/r005/01day/*.nc',
    combine='by_coords',
    chunks={'time':'auto'}
)
print("tmax dataset:")
print(ds_tmax)
```

**Minimum Temperature DS - 1910-2025**

```python
ds_tmin = xr.open_mfdataset(
    f'{base_path}/tmin/mean/r005/01day/*.nc',
    combine='by_coords',
    chunks={'time':'auto'}
)
print("tmin dataset:")
print(ds_tmin)
```

**Daily Precipitation DS - 1900-2025**

```python
ds_precip = xr.open_mfdataset(
    f'{base_path}/precip/total/r005/01day/*.nc',
    combine='by_coords',
    chunks={'time':'auto'}
)
print("Precip dataset:")
print(ds_precip)
```
